 Что мы можем сказать об игре Надаля?

```python
import numpy as np
import pandas as pd


from scipy.stats import norm, chi2_contingency
import statsmodels.api as sm
```

# Таблица 1. Результаты Надаля на глиняных и других кортах

Нулевая гипотеза:
Тип покрытия корта не оказывает влияния на исход матча, а наблюдаемая разница в проценте побед обусловлена случайными колебаниями.

```python
s1 = 1660           # число успехов (побед) на грунте
n1 = 2523           # общее число матчей на грунте
s2 = 3658           # число успехов (побед) на других покрытиях
n2 = 6373           # общее число матчей на других покрытиях
```

Используем proportions_ztest:

```python
z1, p_value1 = sm.stats.proportions_ztest([s1, s2], [n1, n2])
print(['{:.12f}'.format(b) for b in (z1, p_value1)])
```

**Вывод:**

Z-статистика показывает, на сколько стандартных отклонений наблюдаемая разница в винрейте (65.8% на грунте против 57.4% на других покрытиях) отклоняется от нулевого значения (отсутствие различий). Чем больше абсолютное значение Z, тем сильнее выражено различие.

Полученное p-значение значительно меньше порога 0.05, что позволяет отвергнуть нулевую гипотезу. Таким образом, разница в проценте побед Надаля на грунтовых и других кортах является статистически значимой. Это свидетельствует о том, что Надаль демонстрирует более высокую результативность на грунтовом покрытии по сравнению с другими.

# Таблица 2. Результаты Надаля на глиняных и других кортах - геймы, в которых он подавал

Нулевая гипотеза:
Тип покрытия корта не влияет на вероятность выигрыша Надалем гейма на своей подаче.

```python
s1 = 1090
n1 = 1261
s2 = 2747
n2 = 3193
```

```python
z1, p_value1 = sm.stats.proportions_ztest([s1, s2], [n1, n2])
print(['{:.12f}'.format(b) for b in (z1, p_value1)])
```

**Вывод:**

Полученное значение Z-статистики является небольшим по модулю, что указывает на то, что разница в проценте выигранных геймов на своей подаче между грунтом и другими покрытиями незначительна по сравнению с естественной вариативностью данных.

p-значение превышает 0.05, поэтому у нас нет оснований отвергнуть нулевую гипотезу. Следовательно, несмотря на то что Надаль чаще выигрывает матчи на грунте, анализ геймов на его подаче показывает, что покрытие корта не оказывает статистически значимого влияния на эффективность его подачи. Процент выигранных подач остаётся примерно одинаковым на любом покрытии.

# Таблица 3. Результаты Надаля на глиняных и других кортах — геймы, в которых он принимал подачу
Нулевая гипотеза:
Тип покрытия корта не влияет на способность Надаля выигрывать геймы при приёме подачи.

```python
s1 = 570
n1 = 1262
s2 = 911
n2 = 3180

z1, p_value1 = sm.stats.proportions_ztest([s1, s2], [n1, n2])
print(['{:.12f}'.format(b) for b in (z1, p_value1)])
```

**Вывод:**

Z-статистика принимает очень большое по модулю значение, что свидетельствует о колоссальной разнице в эффективности Надаля на приёме подачи в зависимости от покрытия. На грунте он значительно чаще выигрывает геймы на приёме, чем на других покрытиях.

p-значение существенно меньше 0.05, поэтому нулевая гипотеза отвергается. Разница в проценте выигранных геймов на приёме между грунтом и другими покрытиями является статистически значимой.



**Заключительный вывод по всем трём тестам**

Проведённый анализ показывает, что преимущество Надаля на грунте обусловлено не его подачей (которая остаётся стабильно эффективной на любых покрытиях), а игрой на приёме — именно на грунтовых кортах он значительно чаще выигрывает геймы на подаче соперника. Это объясняет его выдающуюся статистику побед на данном типе покрытия.
